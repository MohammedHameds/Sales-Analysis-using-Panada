```python
df.columns = df.columns.str.strip().str.title().str.replace(' ', '_')
```

```python
## Standardization
## Normalization and cleaning
## Data types
## Missing values
```

```python
df['Retailer'].value_counts()
```

```python
df['Retailer'] = df['Retailer'].str.title()
```

```python
df['Retailer'].value_counts()
```

```python
df['Retailer'].isnull().sum()
```

```python
df.drop('Retailer_Id', axis = 1 , inplace = True)
```

```python
df['Invoice_Date'] = df['Invoice_Date'].astype('datetime64[ns]')
```

```python
df['Invoice_Date']
```

```python
df['Region'].value_counts()
```

```python
most_region = df['Region'].mode()[0]
```

```python
df['Region'].isnull().sum()
```

```python
df['Region'] = df['Region'].fillna(most_region)
```

```python
df['Region'].isnull().sum()
```

```python
df['State'].value_counts()
```

```python
df['City'].value_counts()
```

```python
df['City'] = df['City'].str.title()
```

```python
df['City'].value_counts()
```

```python
df['Product'].value_counts()
```

```python
df['Product'] = df['Product'].replace("Men's aparel", "Men's Apparel")
```

```python
df['Product'].value_counts()
```

```python
"Men's Athletic Footwear ".split("'")
```

```python
df['Product'].str.split("'").str[0]
```

```python
df['Price_Per_Unit'] = df['Price_Per_Unit'].str.strip().str.replace('$', '')
df['Price_Per_Unit'] = df['Price_Per_Unit'].astype('float').astype('Int64')
```

```python
df['Price_Per_Unit'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
## Price * Units_Sold  = total sales
df['Total_Sales'] / df['Units_Sold']
```

```python
df['Units_Sold'] = df['Units_Sold'].str.replace(',', '').astype('Int64')
```

```python
df['Total_Sales'] = df['Total_Sales'].str.replace(',', '').astype('Int64')
```

```python
df['Price_Per_Unit'] = df['Total_Sales'] / df['Units_Sold']
df['Price_Per_Unit'] = df['Price_Per_Unit'].astype('Int64')
```

```python
df_test = df['c'].astype('float') / df['Units_Sold'].astype('float')
df_test.isnull().sum()
```

```python
df[df['Total_Sales'] == 0]
```

```python
df.dropna(inplace=True)
```

```python
df['Operating_Profit'] = df['Operating_Profit'].str.replace({'$': '', ',': ''}).astype(int)
```

```python
df['Sales_Method'].value_counts()
```

```python
df['Sales_Method'].isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.columns
```

```python
df['Year'] = df['Invoice_Date'].dt.year
df['Month'] = df['Invoice_Date'].dt.month
df['Day'] = df['Invoice_Date'].dt.day
```

```python
df.to_csv('data_cleaned.csv', index=False)
```
